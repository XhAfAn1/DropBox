```python
class GridEnv:
    def __init__(self, size, goal, agentPos, visited):
        self.size = size
        self.goal = goal
        self.agentPos = agentPos
        self.visited = visited

    def is_goal(self):
        return self.goal == self.agentPos


class Agent007:
    def __init__(self, env):
        self.env = env
        self.agentX = self.env.agentPos[0]
        self.agentY = self.env.agentPos[1]

    def perceive(self, stack):
        data = []

        if (self.agentY - 1 > 0 and
                [self.agentX, self.agentY - 1] not in self.env.visited and
                [self.agentX, self.agentY - 1] not in stack):
            data.append([self.agentX, self.agentY - 1])

        if (self.agentX - 1 > 0 and
                [self.agentX - 1, self.agentY] not in self.env.visited and
                [self.agentX - 1, self.agentY] not in stack):
            data.append([self.agentX - 1, self.agentY])

        if (self.agentY + 1 <= self.env.size and
                [self.agentX, self.agentY + 1] not in self.env.visited and
                [self.agentX, self.agentY + 1] not in stack):
            data.append([self.agentX, self.agentY + 1])

        if (self.agentX + 1 <= self.env.size and
                [self.agentX + 1, self.agentY] not in self.env.visited and
                [self.agentX + 1, self.agentY] not in stack):
            data.append([self.agentX + 1, self.agentY])

        return data


# Environment
g1 = GridEnv(4, [1, 1], [3, 1], [])

# Agent
masudRana = Agent007(g1)

# Stack (LIFO)
stack = []
stack.append(masudRana.env.agentPos)

while stack:
    agent_current_state = stack.pop()  # Pop from top of stack

    masudRana.agentX = agent_current_state[0]
    masudRana.agentY = agent_current_state[1]

    print("Masud rana in", agent_current_state)

    g1.visited.append(agent_current_state)

    if g1.goal == agent_current_state:
        print("goal reached")
        break

    data = masudRana.perceive(stack)

    # Push neighbors onto stack
    for d in data:
        stack.append(d)
```
