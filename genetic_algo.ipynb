```python
import random

def getChrom(size):
    chrom = []
    for i in range(size):
        chrom.append(random.randint(1, size))
    return chrom

def crossOver(chrom1, chrom2, size):
    k = random.randint(2, size - 1)
    c1 = chrom1[:k] + chrom2[k:]
    c2 = chrom2[:k] + chrom1[k:]
    return c1, c2

def mutate(chrom, size):
    k1 = random.randint(0, size - 1)
    k2 = random.randint(1, size)
    chrom[k1] = k2
    return chrom

def fitness(chrom, size):
    total = (size * (size - 1)) // 2

    for i in range(size):
        for j in range(i + 1, size):
            if chrom[i] == chrom[j]:
                total -= 1

    return total
```

```python
size = 8

for generation in range(1, 11):

    print(f"\n{'='*20}")
    print(f"GENERATION {generation}")
    print(f"{'='*20}")

    # Create 6 parents
    parents = []
    for i in range(6):
        chrom = getChrom(size)
        fit = fitness(chrom, size)
        parents.append((chrom, fit))

    # Print parents
    print("\nParents:")
    for i, (chrom, fit) in enumerate(parents):
        print(f"P{i+1}: {chrom}  Fitness = {fit}")

    # Select best 2 parents
    parents.sort(key=lambda x: x[1], reverse=True)

    parent1 = parents[0][0]
    parent2 = parents[1][0]

    print("\nBest Parents:")
    print("Parent 1:", parent1, "Fitness =", parents[0][1])
    print("Parent 2:", parent2, "Fitness =", parents[1][1])

    # Create 6 children
    children = []

    while len(children) < 6:
        c1, c2 = crossOver(parent1, parent2, size)

        c1 = mutate(c1, size)
        c2 = mutate(c2, size)

        children.append(c1)

        if len(children) < 6:
            children.append(c2)

    # Print children and fitness
    print("\nChildren After Crossover + Mutation:")
    for i, child in enumerate(children):
        fit = fitness(child, size)
        print(f"C{i+1}: {child}  Fitness = {fit}")
```

```python
import random
import matplotlib.pyplot as plt

# -----------------------------
# Generate a chromosome
# -----------------------------
def getChrom(size):
    chrom = []
    for i in range(size):
        chrom.append(random.randint(1, size))
    return chrom


# -----------------------------
# Crossover
# -----------------------------
def crossOver(chrom1, chrom2, size):
    k = random.randint(2, size - 1)

    c1 = chrom1[:k] + chrom2[k:]
    c2 = chrom2[:k] + chrom1[k:]

    return c1, c2


# -----------------------------
# Mutation
# -----------------------------
def mutate(chrom, size):
    k1 = random.randint(0, size - 1)
    k2 = random.randint(1, size)

    chrom[k1] = k2

    return chrom


# -----------------------------
# N-Queens Fitness Function
# -----------------------------
def fitness(chrom, size):
    max_fitness = (size * (size - 1)) // 2

    attacks = 0

    for i in range(size):
        for j in range(i + 1, size):

            # Same row
            if chrom[i] == chrom[j]:
                attacks += 1

            # Same diagonal
            elif abs(chrom[i] - chrom[j]) == abs(i - j):
                attacks += 1

    return max_fitness - attacks


# ==================================
# Parameters
# ==================================

size = 8
generations = 10
population_size = 6

best_fitness_history = []

# ==================================
# Initial Population (Generation 1)
# ==================================

population = [getChrom(size) for _ in range(population_size)]

# ==================================
# Evolution Loop
# ==================================

for generation in range(1, generations + 1):

    print("\n" + "=" * 60)
    print(f"GENERATION {generation}")
    print("=" * 60)

    # --------------------------------
    # Evaluate Population
    # --------------------------------

    scored_population = []

    print("\nPopulation:")

    for i, chrom in enumerate(population):

        fit = fitness(chrom, size)

        scored_population.append((chrom, fit))

        print(f"{i+1}. {chrom}   Fitness = {fit}")

    # --------------------------------
    # Sort by Fitness
    # --------------------------------

    scored_population.sort(key=lambda x: x[1], reverse=True)

    parent1 = scored_population[0][0]
    parent2 = scored_population[1][0]

    best_fit = scored_population[0][1]

    best_fitness_history.append(best_fit)

    print("\nBest Parents Selected:")

    print(f"Parent 1: {parent1}")
    print(f"Fitness : {scored_population[0][1]}")

    print(f"\nParent 2: {parent2}")
    print(f"Fitness : {scored_population[1][1]}")

    # Solution found
    max_possible = (size * (size - 1)) // 2

    if best_fit == max_possible:
        print("\nOptimal N-Queens solution found!")
        break

    # --------------------------------
    # Generate Children
    # --------------------------------

    children = []

    while len(children) < population_size:

        c1, c2 = crossOver(parent1, parent2, size)

        c1 = mutate(c1.copy(), size)
        c2 = mutate(c2.copy(), size)

        children.append(c1)

        if len(children) < population_size:
            children.append(c2)

    print("\nChildren:")

    child_scores = []

    for i, child in enumerate(children):

        fit = fitness(child, size)

        child_scores.append((child, fit))

        print(f"C{i+1}: {child}   Fitness = {fit}")

    # --------------------------------
    # Next Generation
    # --------------------------------
    # Population becomes the children
    # The best 2 children will be selected
    # automatically in the next iteration.

    population = children

# ==================================
# Results
# ==================================

print("\n" + "=" * 60)
print("BEST FITNESS PER GENERATION")
print("=" * 60)

for gen, fit in enumerate(best_fitness_history, start=1):
    print(f"Generation {gen}: {fit}")

# ==================================
# Best Solution Found
# ==================================

final_population = [(chrom, fitness(chrom, size)) for chrom in population]
final_population.sort(key=lambda x: x[1], reverse=True)

print("\nBest Chromosome Found:")
print(final_population[0][0])
print("Fitness =", final_population[0][1])

# ==================================
# Plot Improvement
# ==================================

plt.figure(figsize=(10, 5))

plt.plot(
    range(1, len(best_fitness_history) + 1),
    best_fitness_history,
    marker='o',
    linewidth=2
)

plt.title("N-Queens Genetic Algorithm Progress")
plt.xlabel("Generation")
plt.ylabel("Best Fitness")
plt.grid(True)

for x, y in zip(
    range(1, len(best_fitness_history) + 1),
    best_fitness_history
):
    plt.text(x, y, str(y))

plt.show()
```
